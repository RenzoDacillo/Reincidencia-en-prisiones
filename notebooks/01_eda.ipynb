## **Eliminación de features irrelevantes**

Se procede a quitar las columnas redundantes, con descripciones(texto) y las que no influyen considerablemente en la predicción de la reincidencia. Este es un proceso manual y largo ya que cada columna tiene un nombre particular y se debe de justificar con la literatura, aunque también se usará el sentido común como justificación.


Para ello referenciaremos la publicación:
 `Understanding and Predicting Recidivism in Latin America: A Machine Learning Approach `
con el fin de identificar los features que mantendremos o retiraremos. Las referencias están todas sacadas de esta publicación.

Según los autores, las predictores más importantes de la reincidencia son la **edad** y si el cargo fue un **delito económico o financiero**(como hurto y/o tentativa de hurto o crimenes contra la propiedad), por otro lado, los **factores geográficos** juegan un papel considerable ya que los reincidentes suelen concentrarse en ciertas prisiones. Otros factores como la duración de la sentencia, la sansión o las condiciones de la prisión, no son menos relevantes pero aun por tomar en cuenta.
(Anauati et.al,2026)


El género no parece relevante, si bien los hombres cometen más crimenes en general, los predictores de la reincidencia son muy similares entre ambos sexos (Safranoff & Tiravassi,2018; Mears et al., 2012)

Los **antecedentes familiares** y **los factores socioeconómicos** son otros factores que influencian la decisión de cometer un crimen. Esto es esperable ya que la mayoría de zonas inseguras son las más pobres. Desafíos estructurales relacionados con la pobreza
 como **inestabilidad de la vivienda y desventajas en el entorno vecinal** aumentan el riesgo de acciones criminales (Clark, 2016). Investigaciones mencionaron que individuos que experimentaron **experiencias adversas en su niñez** son más propensos a cometer un crimen posteriormente (Graf et al., 2021; Basto-Pereira et al., 2022; Reavis et al., 2013).

Lo que sucede dentro de las prisiones también afecta este riesgo, no solo la **duración de la sentencia**, sino la calidad en la que viven los reclusos. Factores como la sobrepoblación, aislamiento, violencia, falta de acceso a programas de rehabilitación dentro de la prisión pueden aumentar el riesgo. Por ello, la **localización geográfica de la prisión** juega un papel importante en la predicción.

Entonces, tanto los factores personales, institucionales y sociales luego de la liberación del recluso influyen.

```python
#Código sacado del notebook de clase (introducción al preprocesamiento de datos)
# Calcular IQR
columnas_numericas = Dataset.select_dtypes(include=['number'])
Q1 = columnas_numericas.iloc[:, :-1].quantile(0.25)
Q3 = columnas_numericas.iloc[:, :-1].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Outliers por columna (booleano)
outliers_iqr = (columnas_numericas.iloc[:, :-1] < lower_bound) | (columnas_numericas.iloc[:, :-1] > upper_bound)

# Resumen
outliers_summary = pd.DataFrame({
    'Columna': outliers_iqr.columns,
    'Outliers': outliers_iqr.sum().values,
    'Porcentaje (%)': (outliers_iqr.sum().values / len(columnas_numericas) * 100).round(2)
})

outliers_summary.sort_values('Outliers', ascending=False)
```

Se eliminarán estos outliers sin más, dado que si bien es una minoría, la cantidad absoluta de casos es considerable (2445, 1983 y 367 casos) como para analizar una por una.
