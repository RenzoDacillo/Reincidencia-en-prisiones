### **Análisis Chi-cuadrado**

Luego de la selección inicial, se realiza un análisis de chi-cuadrado para ver la relación entre cada columna con el target de manera individual. Se realiza este análisis ya que la mayoría de las categorías son cualitativas y no es posible realizar una matriz de correlación.

```python
# --- CONFIGURACIÓN INICIAL ---
# Sustituye 'reincidencia' por el nombre exacto de tu columna principal
variable_principal = 'REINCIDENTE'

# Lista de respuestas ambiguas que deseas excluir del análisis para no alterar el Chi-cuadrado
valores_a_excluir = ['No sabe / No contesta', 'No sabe', 'No contesta', 'NS/NC', 'Sin dato']

# Creamos una lista para almacenar los resultados de cada variable
resultados_chi = []

print("--- INICIANDO ANÁLISIS DE CHI-CUADRADO ---")

# --- BUCLE DE ANÁLISIS AUTOMÁTICO ---
for columna in Dataset1.columns:
    # Evitamos cruzar la variable principal con ella misma
    if columna == variable_principal:
        continue

    try:
        # 1. Limpieza dinámica: Filtramos filas con datos perdidos o respuestas ambiguas en ambas columnas
        df_filtrado = Dataset1[
            Dataset1[columna].notna() &
            Dataset1[variable_principal].notna() &
            (~Dataset1[columna].isin(valores_a_excluir)) &
            (~Dataset1[variable_principal].isin(valores_a_excluir))
        ]

        # Guardamos la cantidad de filas válidas que quedaron para esta prueba
        n_casos = len(df_filtrado)

        # Seguridad: Si tras limpiar quedan muy pocos datos o categorías, saltamos la columna
        if df_filtrado[columna].nunique() < 2 or df_filtrado[variable_principal].nunique() < 2:
            print(f"⚠️ Omitida: {columna} (Falta de variabilidad o categorías tras limpiar)")
            continue

        # 2. Construcción de la tabla de contingencia
        tabla_contingencia = pd.crosstab(df_filtrado[columna], df_filtrado[variable_principal])

        # 3. Cálculo de Chi-cuadrado
        chi2, p_valor, dof, esperados = stats.chi2_contingency(tabla_contingencia)

        # 4. Interpretación de la significancia estadística (Umbral clásico de 0.05)
        significativo = "Sí" if p_valor < 0.05 else "No"

        # Guardamos las métricas en nuestra lista
        resultados_chi.append({
            'Variable Predictora': columna,
            'Chi-Cuadrado (χ²)': round(chi2, 2),
            'Valor p': round(p_valor, 4) if p_valor >= 0.0001 else "< 0.0001",
            'Grados de Libertad': dof,
            'Casos Válidos': n_casos,
            '¿Es Significativa?': significativo
        })

    except Exception as e:
        print(f"❌ Error al procesar la columna '{columna}': {e}")

# --- PRESENTACIÓN DE RESULTADOS ---
# Convertimos los resultados a un nuevo DataFrame para manipularlos fácilmente
df_resultados = pd.DataFrame(resultados_chi)

# Ordenamos la tabla de mayor a menor impacto (las variables con Chi-cuadrado más alto primero)
df_resultados = df_resultados.sort_values(by='Chi-Cuadrado (χ²)', ascending=False).reset_index(drop=True)

print("\n--- TABLA RESUMEN DE ASOCIACIONES CON LA REINCIDENCIA ---")
print(df_resultados.to_string())

# Opcional: Guarda este reporte directo a un archivo Excel para tu tesis o informe
# df_resultados.to_excel('resultados_chi_cuadrado.xlsx', index=False)
```

```python
#Visualización de las categorías con mayor relación con el Target

variable_principal = 'REINCIDENTE'
valores_a_excluir = ['No sabe / No contesta', 'No sabe', 'No contesta', 'NS/NC', 'Sin dato']
resultados_chi = []

# --- BUCLE DE ANÁLISIS AUTOMÁTICO ---
for columna in Dataset1.columns:
    if columna == variable_principal:
        continue

    try:
        df_filtrado = Dataset1[
            Dataset1[columna].notna() &
            Dataset1[variable_principal].notna() &
            (~Dataset1[columna].isin(valores_a_excluir)) &
            (~Dataset1[variable_principal].isin(valores_a_excluir))
        ]

        if df_filtrado[columna].nunique() < 2 or df_filtrado[variable_principal].nunique() < 2:
            continue

        tabla_contingencia = pd.crosstab(df_filtrado[columna], df_filtrado[variable_principal])
        chi2, p_valor, dof, esperados = stats.chi2_contingency(tabla_contingencia)

        # Guardamos solo las que tengan una relación estadísticamente real (p < 0.05)
        if p_valor < 0.05:
            resultados_chi.append({
                'Variable': columna,
                'Chi-Cuadrado': chi2,
                'Valor p': p_valor
            })

    except Exception as e:
        pass

# --- PROCESAMIENTO DE LOS TOP 10 ---
df_resultados = pd.DataFrame(resultados_chi)

# Ordenamos de mayor a menor impacto y tomamos los 10 primeros
df_top10 = df_resultados.sort_values(by='Chi-Cuadrado', ascending=False).head(10)

# --- GENERACIÓN DEL GRÁFICO ---
plt.figure(figsize=(10, 6))

# Creamos un gráfico de barras horizontales usando Seaborn
# El color coral degradado resalta visualmente el peso de las variables
ax = sns.barplot(
    x='Chi-Cuadrado',
    y='Variable',
    data=df_top10,
    palette='Blues',
    edgecolor='black'
)

# Añadimos las etiquetas del valor de Chi-cuadrado al final de cada barra
for p in ax.patches:
    width = p.get_width()
    ax.text(
        width + (width * 0.01), # Posición horizontal de la etiqueta
        p.get_y() + p.get_height() / 2, # Posición vertical (centrado)
        f'{width:.1f}', # Formato con un decimal
        va='center',
        ha='left',
        fontsize=10,
        fontweight='bold'
    )

# Personalización del diseño del gráfico para presentación o tesis
plt.title('Top 10 Variables con Mayor Relación Estadística con la Reincidencia', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Fuerza de la Relación (Valor del Estadístico Chi-Cuadrado)', fontsize=11, labelpad=10)
plt.ylabel('Variables Predictoras', fontsize=11)

# Limpiamos el borde superior y derecho del gráfico para una estética más moderna
sns.despine(top=True, right=True)

plt.tight_layout()
plt.show()
```

###Outliers cuantitativos


```python
# Calcular IQR
columnas_numericas = Dataset[['P124','P122','EDAD']].copy()
Q1 = columnas_numericas.iloc[:, :-1].quantile(0.25)
Q3 = columnas_numericas.iloc[:, :-1].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Outliers por columna (booleano)
outliers_iqr = (columnas_numericas.iloc[:, :-1] < lower_bound) | (columnas_numericas.iloc[:, :-1] > upper_bound)

# Resumen
outliers_summary = pd.DataFrame({
    'Columna': outliers_iqr.columns,
    'Outliers': outliers_iqr.sum().values,
    'Porcentaje (%)': (outliers_iqr.sum().values / len(columnas_numericas) * 100).round(2)
})

outliers_summary.sort_values('Outliers', ascending=False)
```
